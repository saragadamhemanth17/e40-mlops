```python
import random
import csv

random.seed(42)

# Number of original students
num_students = 1000

departments = ["CSE", "AI&DS", "IT", "ECE", "EEE", "Mechanical"]
genders = ["Male", "Female"]

dataset = []

header = [
    "Student_ID",
    "Age",
    "Gender",
    "Department",
    "Semester",
    "CGPA",
    "Academic_Score",
    "Attendance_Percentage",
    "Communication_Skill",
    "Programming_Skill",
    "Problem_Solving",
    "Aptitude_Score",
    "Logical_Reasoning",
    "Technical_Test_Score",
    "Coding_Assessment",
    "Industry_Skill",
    "Internship_Count",
    "Internship_Performance",
    "Projects_Completed",
    "Mini_Projects",
    "Major_Project_Score",
    "Certifications",
    "Hackathon_Participation",
    "Workshops_Attended",
    "Technical_Events",
    "Soft_Skill_Rating",
    "Resume_Score",
    "Interview_Score",
    "Skill_Gap",
    "Employability_Percentage",
    "Placement_Package_LPA",
    "Employment_Status"
]

# -----------------------------
# Generate Dataset
# -----------------------------
for i in range(1, num_students + 1):

    age = random.randint(18, 24)
    gender = random.choice(genders)
    department = random.choice(departments)
    semester = random.randint(1, 8)

    cgpa = round(random.uniform(5.5, 10.0), 2)
    academic = random.randint(45, 100)
    attendance = random.randint(60, 100)

    communication = random.randint(40, 100)
    programming = random.randint(35, 100)
    problem_solving = random.randint(35, 100)
    aptitude = random.randint(40, 100)
    logical = random.randint(40, 100)
    technical = random.randint(35, 100)
    coding = random.randint(35, 100)
    industry = random.randint(35, 100)

    internship_count = random.randint(0, 4)
    internship_performance = random.randint(40, 100)

    projects = random.randint(1, 8)
    mini_projects = random.randint(1, 6)
    major_project = random.randint(50, 100)

    certifications = random.randint(0, 10)
    hackathons = random.randint(0, 8)
    workshops = random.randint(0, 12)
    technical_events = random.randint(0, 10)

    soft_skills = random.randint(40, 100)
    resume = random.randint(40, 100)
    interview = random.randint(35, 100)

    gap = abs(academic - industry)

    employability = (
        0.20 * academic +
        0.20 * industry +
        0.10 * programming +
        0.10 * communication +
        0.10 * aptitude +
        0.10 * interview +
        0.05 * resume +
        0.05 * soft_skills +
        0.05 * technical +
        0.05 * coding
    )

    employability += certifications * 0.5
    employability += internship_count * 1.5
    employability += projects * 0.5
    employability -= gap * 0.2

    employability = round(max(0, min(100, employability)), 2)

    if employability >= 85:
        package = round(random.uniform(8, 18), 2)
    elif employability >= 75:
        package = round(random.uniform(5, 8), 2)
    elif employability >= 65:
        package = round(random.uniform(3, 5), 2)
    else:
        package = 0.0

    status = "Employed" if employability >= 70 else "Not Employed"

    row = [
        f"S{i:03}",
        age,
        gender,
        department,
        semester,
        cgpa,
        academic,
        attendance,
        communication,
        programming,
        problem_solving,
        aptitude,
        logical,
        technical,
        coding,
        industry,
        internship_count,
        internship_performance,
        projects,
        mini_projects,
        major_project,
        certifications,
        hackathons,
        workshops,
        technical_events,
        soft_skills,
        resume,
        interview,
        gap,
        employability,
        package,
        status
    ]

    dataset.append(row)

# -----------------------------
# Add Missing Values (5%)
# -----------------------------
missing_percentage = 0.05

for row in dataset:

    num_missing = random.randint(1, 3)

    for _ in range(num_missing):

        if random.random() < missing_percentage:

            col = random.randint(1, len(header)-2)

            row[col] = ""

# -----------------------------
# Add Duplicate Rows (3%)
# -----------------------------
duplicates = random.sample(dataset, int(0.03 * num_students))

dataset.extend([row.copy() for row in duplicates])

# -----------------------------
# Shuffle Dataset
# -----------------------------
random.shuffle(dataset)

# -----------------------------
# Save CSV
# -----------------------------
with open("student_curriculum_dataset.csv", "w", newline="") as file:

    writer = csv.writer(file)

    writer.writerow(header)

    writer.writerows(dataset)

# -----------------------------
# Print Dataset
# -----------------------------
print(",".join(header))

for row in dataset:
    print(",".join(map(str, row)))

print("\nDataset generated successfully!")

print(f"Original Records : {num_students}")
print(f"Duplicate Records Added : {len(duplicates)}")
print(f"Total Records : {len(dataset)}")

print("\nContains:")
print("✓ Missing Values (~5%)")
print("✓ Duplicate Records (~3%)")
print("✓ 32 Attributes")

print("\nFile saved as student_curriculum_dataset.csv")
```

## 1. Load Dataset

First, we'll load the generated `student_curriculum_dataset.csv` into a pandas DataFrame. We'll then display its head and basic information to get an initial overview of the data, including data types and non-null counts.

```python
import pandas as pd

# Load the dataset
df = pd.read_csv('student_curriculum_dataset.csv')

# Display the first 5 rows
print("First 5 rows of the dataset:")
display(df.head())

# Display basic information about the dataset
print("\nDataset Information:")
display(df.info())
```

```python
# Check for duplicates
print(f"Shape before removing duplicates: {df.shape}")
num_duplicates = df.duplicated().sum()
print(f"Number of duplicate rows found: {num_duplicates}")

# Remove duplicate rows
df.drop_duplicates(inplace=True)
print(f"Shape after removing duplicates: {df.shape}")
```

```python
# Identify numerical and categorical columns
numerical_cols = df.select_dtypes(include=['number']).columns
categorical_cols = df.select_dtypes(include=['object']).columns

# Impute missing values in numerical columns with the mean
for col in numerical_cols:
    if df[col].isnull().any():
        df[col] = pd.to_numeric(df[col], errors='coerce') # Ensure numeric type for mean calculation
        df[col].fillna(df[col].mean(), inplace=True)

# Impute missing values in categorical columns with the mode
for col in categorical_cols:
    if df[col].isnull().any():
        df[col].fillna(df[col].mode()[0], inplace=True)

# Verify that there are no more missing values
print("\nMissing values after imputation:")
display(df.isnull().sum().sum())
```

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Initialize LabelEncoder for the target variable
le = LabelEncoder()
df['Employment_Status_Encoded'] = le.fit_transform(df['Employment_Status'])

# Display the mapping of the target variable
print("Employment Status Mapping:")
for i, status in enumerate(le.classes_):
    print(f"{status}: {i}")

# Identify categorical features for one-hot encoding (excluding the target and Student_ID)
categorical_features = ['Gender', 'Department']

# Create a column transformer for one-hot encoding
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ], remainder='passthrough'
)

# Apply the transformations
# IMPORTANT: Drop 'Employability_Percentage' and 'Placement_Package_LPA' to prevent data leakage
X = df.drop(['Student_ID', 'Employment_Status', 'Employment_Status_Encoded', 'Employability_Percentage', 'Placement_Package_LPA'], axis=1)
y = df['Employment_Status_Encoded']

X_processed = preprocessor.fit_transform(X)

# Get feature names after one-hot encoding
onehot_feature_names = preprocessor.named_transformers_['cat'].get_feature_names_out(categorical_features)
remaining_feature_names = [col for col in X.columns if col not in categorical_features]
all_feature_names = list(onehot_feature_names) + list(remaining_feature_names)

# Convert to DataFrame for better readability (optional, but good for inspection)
X_df = pd.DataFrame(X_processed, columns=all_feature_names)

print("\nFeatures after one-hot encoding (first 5 rows):")
display(X_df.head())
```

```python
from sklearn.preprocessing import StandardScaler

# Initialize StandardScaler
scaler = StandardScaler()

# Apply scaling to the processed features
X_scaled = scaler.fit_transform(X_df)

# Convert back to DataFrame for better readability (optional)
X_scaled_df = pd.DataFrame(X_scaled, columns=X_df.columns)

print("Features after scaling (first 5 rows):")
display(X_scaled_df.head())
```

```python
from sklearn.model_selection import train_test_split

# Split the scaled data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_scaled_df, y, test_size=0.2, random_state=42)

print(f"Training features shape: {X_train.shape}")
print(f"Testing features shape: {X_test.shape}")
print(f"Training target shape: {y_train.shape}")
print(f"Testing target shape: {y_test.shape}")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

# Initialize the models
models = {
    'Logistic Regression': LogisticRegression(random_state=42, solver='liblinear'),
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'Random Forest': RandomForestClassifier(random_state=42, n_estimators=100)
}

# Train each model
print("Training models...")
for name, model in models.items():
    model.fit(X_train, y_train)
    print(f"{name} trained successfully.")
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

results = {}

print("\nEvaluating models...")
for name, model in models.items():
    y_pred = model.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    report = classification_report(y_test, y_pred, target_names=le.classes_)
    cm = confusion_matrix(y_test, y_pred)

    results[name] = {
        'accuracy': accuracy,
        'report': report,
        'confusion_matrix': cm
    }

    print(f"\n--- {name} ---")
    print(f"Accuracy: {accuracy:.4f}")
    print("\nClassification Report:")
    print(report)

    print("\nConfusion Matrix:")
    plt.figure(figsize=(4, 3))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=le.classes_, yticklabels=le.classes_)
    plt.title(f'Confusion Matrix for {name}')
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.show()
```

```python
print("\n--- Model Comparison ---")
best_accuracy = 0
best_model_name = ""

for name, metrics in results.items():
    print(f"{name} Accuracy: {metrics['accuracy']:.4f}")
    if metrics['accuracy'] > best_accuracy:
        best_accuracy = metrics['accuracy']
        best_model_name = name

print(f"\nConclusion: The best performing model based on accuracy is {best_model_name} with an accuracy of {best_accuracy:.4f}.")
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier

# Define the parameter grid to search
param_grid = {
    'n_estimators': [100, 200, 300],  # Number of trees in the forest
    'max_depth': [None, 10, 20, 30],  # Maximum depth of the tree
    'min_samples_split': [2, 5, 10],  # Minimum number of samples required to split an internal node
    'min_samples_leaf': [1, 2, 4]    # Minimum number of samples required to be at a leaf node
}

# Initialize the GridSearchCV object
# Using the best hyperparameters for Random Forest from previous run in the original notebook.
grid_search = GridSearchCV(
    estimator=RandomForestClassifier(random_state=42),
    param_grid=param_grid,
    cv=5,
    n_jobs=-1, # Use all available cores
    verbose=1, # Show progress
    scoring='accuracy'
)

print("Starting GridSearchCV for Random Forest...")
# Fit GridSearchCV to the training data
grid_search.fit(X_train, y_train)

print("GridSearchCV complete.")
print(f"Best parameters found: {grid_search.best_params_}")
print(f"Best cross-validation accuracy: {grid_search.best_score_:.4f}")

# Get the best model
best_rf_model = grid_search.best_estimator_
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# Make predictions with the best Random Forest model
y_pred_tuned_rf = best_rf_model.predict(X_test)

# Evaluate the tuned model
accuracy_tuned_rf = accuracy_score(y_test, y_pred_tuned_rf)
report_tuned_rf = classification_report(y_test, y_pred_tuned_rf, target_names=le.classes_)
cm_tuned_rf = confusion_matrix(y_test, y_pred_tuned_rf)

print("\n--- Tuned Random Forest --- ")
print(f"Accuracy: {accuracy_tuned_rf:.4f}")
print("\nClassification Report:")
print(report_tuned_rf)

print("\nConfusion Matrix:")
plt.figure(figsize=(4, 3))
sns.heatmap(cm_tuned_rf, annot=True, fmt='d', cmap='Blues', xticklabels=le.classes_, yticklabels=le.classes_)
plt.title('Confusion Matrix for Tuned Random Forest')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()

# Update results with the tuned Random Forest model
results['Tuned Random Forest'] = {
    'accuracy': accuracy_tuned_rf,
    'report': report_tuned_rf,
    'confusion_matrix': cm_tuned_rf
}

print("\n--- Updated Model Comparison ---")
best_accuracy = 0
best_model_name = ""

for name, metrics in results.items():
    print(f"{name} Accuracy: {metrics['accuracy']:.4f}")
    if metrics['accuracy'] > best_accuracy:
        best_accuracy = metrics['accuracy']
        best_model_name = name

print(f"\nConclusion: The best performing model based on accuracy is {best_model_name} with an accuracy of {best_accuracy:.4f}.")
```
